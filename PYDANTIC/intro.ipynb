```python
from pydantic import BaseModel
from typing import Optional

class Student(BaseModel):
    name: str
    age: int
    branch: str
    phone: Optional[str] = None
    linkedin: Optional[str] = None
```

```python
# Creating object without optional fields
student1 = Student(
    name="Aryan",
    age=20,
    branch="CSE AIML"
)

print(student1)

# Creating object with optional fields
student2 = Student(
    name="Rahul",
    age=21,
    branch="CSE",
    phone="9876543210",
    linkedin="linkedin.com/in/rahul"
)

print(student2)
```

```python
from pydantic import BaseModel


class Address(BaseModel):
    city: str
    state: str
    pincode: int


class Academic(BaseModel):
    branch: str
    semester: int
    cgpa: float


class Student(BaseModel):
    name: str
    age: int
    address: Address
    academic: Academic


student = Student(
    name="Aryan",
    age=20,

    address={
        "city": "Delhi",
        "state": "Delhi",
        "pincode": 110085
    },

    academic={
        "branch": "CSE AIML",
        "semester": 3,
        "cgpa": 9.09
    }
)


print(student)
```

```python
print(student.name)

print(student.address.city)

print(student.address.pincode)

print(student.academic.branch)

print(student.academic.cgpa)
```

```python
print(student.model_dump())

print(student.model_dump_json(indent=4))
```

```python
from pydantic import BaseModel, Field

class Product(BaseModel):
    name: str
    price: float = Field(gt=0)
    quantity: int = Field(ge=1, le=100)

product = Product(
    name="Laptop",
    price=55000,
    quantity=2
)

print(product)
```

```python
class User(BaseModel):
    username: str = Field(
        min_length=3,
        max_length=15,
        description="Username must contain 3-15 characters"
    )

user = User(username="Aryan123")
print(user)
```

```python
class Employee(BaseModel):
    name: str
    department: str = Field(default="IT")
    salary: float = Field(default=25000, gt=0)

employee = Employee(name="Aryan")

print(employee)
```

```python
from pydantic import BaseModel, Field

class TravelPlan(BaseModel):

    destination: str = Field(
        description="Destination city or country"
    )

    days: int = Field(
        gt=0,
        le=30,
        description="Number of travel days"
    )

    budget: float = Field(
        gt=0,
        description="Total budget in INR"
    )

    activities: list[str] = Field(
        min_length=1,
        description="List of planned activities"
    )
```
