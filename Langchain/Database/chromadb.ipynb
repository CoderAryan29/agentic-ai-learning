```python
from langchain_ollama import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
embeddings = OllamaEmbeddings(
    model="nomic-embed-text"
)
```

```python
vector = embeddings.embed_query(
    "I am learning Agentic AI"
)

print(len(vector))
print(vector[:5])
```

```python
documents = [
    Document(
        page_content="Farmers can apply for government subsidies through official agricultural portals.",
        metadata={"category": "farming"}
    ),
    Document(
        page_content="Artificial intelligence helps computers learn patterns from data.",
        metadata={"category": "AI"}
    ),
    Document(
        page_content="ChromaDB stores embeddings and supports similarity search.",
        metadata={"category": "database"}
    ),
    Document(
        page_content="Organic farming uses natural methods to improve soil health.",
        metadata={"category": "farming"}
    )
]
```

```python
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embeddings,
    collection_name="ollama_chroma_demo",
    persist_directory="E:/AGENTIC AI/chroma_db"
)
```

```python
results = vectorstore.similarity_search(
    "How can farmers get financial help?",
    k=2
)

for doc in results:
    print(doc.page_content)
    print(doc.metadata)
```

```python
results = vectorstore.similarity_search(
    "How can farmers improve their work?",
    k=2,
    filter={"category": "farming"}
)

for doc in results:
    print(doc.page_content)
```

```python
db = Chroma(
    collection_name="ollama_chroma_demo",
    embedding_function=embeddings,
    persist_directory="E:/AGENTIC AI/chroma_db"
)
```

```python
db.similarity_search(
    "What is ChromaDB?",
    k=2
)
```
