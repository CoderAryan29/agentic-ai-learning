```python
#READING A PDF FILE
from langchain_community.document_loaders import PyPDFLoader
loader=PyPDFLoader("work.pdf")
docs=loader.load()
docs
```

RECURSIVE CHARACTER TEXT SPLITTER

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=5)
text_splitter
```

```python
final_docs=text_splitter.split_documents(docs)
print(final_docs[0])
print(final_docs[1])
```

JSON TEXT SPLITTER

```python
data = {
    "company": "OpenAI",
    "industry": "Artificial Intelligence",
    "employees": [
        {
            "name": "John",
            "role": "Developer",
            "skills": ["Python", "LangChain"]
        },
        {
            "name": "David",
            "role": "AI Engineer",
            "skills": ["Machine Learning", "NLP"]
        },
        {
            "name": "Alex",
            "role": "Data Scientist",
            "skills": ["Pandas", "NumPy"]
        }
    ],
    "technologies": {
        "AI": ["Machine Learning", "Deep Learning"],
        "Frameworks": ["LangChain", "TensorFlow"]
    }
}
```

```python
from langchain_text_splitters import RecursiveJsonSplitter
import json
```

```python
json_splitter=RecursiveJsonSplitter(max_chunk_size=100)
```

```python
json_chunks=json_splitter.split_json(data)
```

```python
for chunk in json_chunks[:3]:
    print(chunk)
```

```python
# can also output documents
docs=json_splitter.create_documents(texts=[data])
for doc in docs[:3]:
    print(doc)
```
